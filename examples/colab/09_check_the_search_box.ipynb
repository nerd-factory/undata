## Run the use case

```python
import undata as un
```

```python
for question in ["child mortality in Bengal", "child mortality in Bangladesh"]:
    answer = un.ask(question)
    print(f"{question!r}: answered about {answer.place}; guessed = {answer.place_was_guessed}")
```
