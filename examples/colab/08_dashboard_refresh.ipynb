# 08 · A dashboard that refreshes itself

> "Our Power BI dashboard needs fresh numbers every month, without anyone remembering to update it."

Refresh the numbers behind a dashboard. Schedule this file; nobody has to run it.

Run each cell in order with **Shift + Enter** (or the ▶ button). You don't need to know Python: change the names in quotes to ask your own question. The page shows what this code returned on its last run: [use case 08](https://nerd-factory.github.io/undata/#dashboard).

It uses [undata](https://nerd-factory.github.io/undata/), a free, open-source package for the [UN System Data Commons](https://data.un.org), which holds statistics from 26 UN System entities. undata is a personal open-source project, not an official UN product, and the data belongs to the agencies that publish it.

## Run the use case

```python
from datetime import date
```

```python
import undata as un
```

```python
INDICATORS = [
    "undata/sdg/EG_ACS_ELEC",
    "undata/unicef/CME_MR.AGE--Y0T4",
    "undata/sdg/SG_GEN_PARL.SEX--F",
]
```

```python
table = un.get(INDICATORS, "Southern Asia", years=(2015, date.today().year))
un.save(table, "dashboard-data.csv")  # Power BI, Tableau or Excel reads this file
```

```python
print(
    f"{date.today()}: wrote {len(table)} rows for {table['country'].nunique()} countries, "
    f"{table['year'].min()} to {table['year'].max()}, to dashboard-data.csv"
)
```
