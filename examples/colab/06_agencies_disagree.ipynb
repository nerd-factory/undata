# 06 · When agencies disagree

> "Two reports quote different maternal mortality figures for Bangladesh. Which is right, and what should we cite?"

One statistic, three agencies. Quote one figure, and name whose it is.

Run each cell in order with **Shift + Enter** (or the ▶ button). You don't need to know Python: change the names in quotes to ask your own question. The page shows what this code returned on its last run: [use case 06](https://nerd-factory.github.io/undata/#sources).

It uses [undata](https://nerd-factory.github.io/undata/), a free, open-source package for the [UN System Data Commons](https://data.un.org), which holds statistics from 26 UN System entities. undata is a personal open-source project, not an official UN product, and the data belongs to the agencies that publish it.

## Run the use case

```python
import undata as un
```

```python
MATERNAL_MORTALITY_RATIO = [
    "undata/sdg/SH_STA_MORT.SEX--F",  # Global SDG Indicators Database
    "undata/unicef/MNCH_MMR.SEX--F",  # UNICEF
    "undata/undphdro/GII_mmr.SEX--F",  # UNDP Human Development Report Office
]
table = un.latest(MATERNAL_MORTALITY_RATIO, "Bangladesh")
print(table[["source", "year", "value"]].round(1).to_string(index=False), "\n")
```

```python
low, high = table["value"].min(), table["value"].max()
odd = table.loc[table["value"].idxmax(), "source"]
print(
    f"For Bangladesh in {table['year'].max()}, the three sources give {low:.0f} to {high:.0f} "
    f"maternal deaths per 100,000 live births; {odd} is the outlier. A gap of "
    f"{(high / low - 1) * 100:.0f}% is enough to change a headline, so cite the source."
)
```
