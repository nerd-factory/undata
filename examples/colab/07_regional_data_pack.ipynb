# 07 · A regional data pack in Excel

> "Send the publications team an Excel file for South-Eastern Asia: electricity, drinking water and primary completion, with sources."

A data pack for the publications team: three indicators, one Excel file, sources included.

Run each cell in order with **Shift + Enter** (or the ▶ button). You don't need to know Python: change the names in quotes to ask your own question. The page shows what this code returned on its last run: [use case 07](https://nerd-factory.github.io/undata/#data-pack).

It uses [undata](https://nerd-factory.github.io/undata/), a free, open-source package for the [UN System Data Commons](https://data.un.org), which holds statistics from 26 UN System entities. undata is a personal open-source project, not an official UN product, and the data belongs to the agencies that publish it.

## Run the use case

```python
import pandas as pd
```

```python
import undata as un
```

```python
INDICATORS = [
    "undata/sdg/EG_ACS_ELEC",  # access to electricity
    "undata/sdg/SH_H2O_SAFE",  # safely managed drinking water
    "undata/sdg/SE_TOT_CPLR.EDUCATION_LEVEL--ISCED11_1",  # primary completion
]
```

```python
table = un.latest(INDICATORS, "South-Eastern Asia")
path = un.save(table, "south-eastern-asia-data-pack.xlsx")
```

```python
sheets = pd.ExcelFile(path).sheet_names
print(f"Saved {len(table)} values for {table['country'].nunique()} countries to {path}")
print("Sheets:", ", ".join(sheets), "\n")
print(un.cite(table))
```
